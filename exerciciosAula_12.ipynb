```python
import numpy as np
from sklearn.linear_model import LinearRegression

# Dados de exemplo
# Quantidade de veículos
x_veiculos = np.array([[1], [2], [3], [4], [5]])

# Quantidade de pessoas envolvidas
y_pessoas = np.array([2, 4, 6, 9, 11])

# Criando o modelo
modelo = LinearRegression()

# Treinando
modelo.fit(x_veiculos, y_pessoas)

# Novo acidente com 6 veículos
veiculos_novos = np.array([[6]])

# Previsão
previsao = modelo.predict(veiculos_novos)

# Resultado
print(f"Para um acidente com {veiculos_novos[0][0]} veículos,")
print(f"a previsão é de aproximadamente {previsao[0]:.0f} pessoas envolvidas.")
```

```python
import numpy as np
from sklearn.linear_model import LinearRegression

# Dados de exemplo
# Total de pessoas envolvidas
x_pessoas = np.array([[2], [4], [6], [8], [10]])

# Quantidade de ilesos
y_ilesos = np.array([1, 2, 3, 4, 5])

# Criando o modelo
modelo = LinearRegression()

# Treinando o modelo
modelo.fit(x_pessoas, y_ilesos)

# Novo acidente com 10 pessoas
pessoas_novas = np.array([[10]])

# Fazendo a previsão
previsao = modelo.predict(pessoas_novas)

# Resultado
print(f"Em um acidente com {pessoas_novas[0][0]} pessoas envolvidas,")
print(f"a previsão é de aproximadamente {previsao[0]:.0f} pessoas ilesas.")
```

```python
import numpy as np
from sklearn.linear_model import LinearRegression

# Dados de exemplo
# Total de feridos
x_feridos = np.array([[2], [4], [6], [8], [10]])

# Quantidade de feridos graves
y_graves = np.array([0, 1, 2, 3, 4])

# Criando o modelo
modelo = LinearRegression()

# Treinando o modelo
modelo.fit(x_feridos, y_graves)

# Novo acidente com 5 feridos
feridos_novos = np.array([[5]])

# Previsão
previsao = modelo.predict(feridos_novos)

# Resultado
print(f"Em um acidente com {feridos_novos[0][0]} feridos,")
print(f"a previsão é de aproximadamente {previsao[0]:.0f} feridos graves.")
```
